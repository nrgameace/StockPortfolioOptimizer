```python
import yfinance as yf
import pandas as pd
import time
```

```python
import os

# Assume the project root is the parent of src/
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
data_dir = os.path.join(project_root, "data", "raw")
os.makedirs(data_dir, exist_ok=True)

print("Saving files to:", data_dir)
```

```python
tickers = ["GOOG", "AAPL", "PLTR", "MSFT", "RKLB"]
ticker_info = yf.download(tickers = tickers[0], start = "2024-01-01", end = "2025-10-01", interval = "1d")[["Adj Close"]].rename(columns={"Adj Close": tickers[0]})
csv_file = os.path.join(data_dir, "all_tickers.csv")


ticker_info.columns = [tickers[0]]
for ticker in tickers[1:]:
        ticker_info.to_csv(csv_file)
        time.sleep(5)
        ticker_info = pd.concat([ticker_info, yf.download(tickers = ticker, start = "2020-01-01", end = "2025-10-01", interval = "1d")], axis=1)
ticker_info
ticker_info.to_csv(csv_file)
```
