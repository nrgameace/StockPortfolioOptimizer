```python
import json
import pandas as pd
from urllib.request import urlopen
import certifi

def get_jsonparsed_data(url):
    response = urlopen(url, cafile=certifi.where())
    data = response.read().decode("utf-8")
    return json.loads(data)

api_key = "YOUR_API_KEY"
tickers = ["AAPL", "GOOG", "MSFT"]
all_data = pd.DataFrame()

for ticker in tickers:
    url = f"https://financialmodelingprep.com/stable/historical-price-eod/dividend-adjusted?symbol={ticker}&apikey={api_key}"
    data = get_jsonparsed_data(url)
    
    if "historical" not in data:
        print(f"No historical data found for {ticker}. Response: {data}")
        continue
    
    df = pd.DataFrame(data['historical'])[['date', 'adjClose']].rename(columns={'adjClose': ticker})
    df.set_index('date', inplace=True)
    
    if all_data.empty:
        all_data = df
    else:
        all_data = all_data.join(df, how='outer')

all_data.sort_index(inplace=True)
all_data.to_csv("all_tickers_adjusted_close.csv")
print(all_data.head())
```
