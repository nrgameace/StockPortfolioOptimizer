```python
tickers = ["APPL", "AMD", "MSFT", "NVDA"]
ticker_data = []

for ticker in tickers:
    ticker_data.append(pd.read_csv(os.path.join(data_dir,f"{ticker}.csv")))

combined = ticker_data[0]

for df in ticker_data[1:]:
   combined = pd.concat([combined,df],axis=1)


combined = combined.drop(["High","Low","Open","Volume"],axis=1)

cols_to_remove = list(range(2,2*len(tickers),2))

combined = combined.iloc[:, [j for j in range(len(combined.columns)) if j not in cols_to_remove]]
path = os.path.join(data_dir,"COMBINED.csv")
combined.to_csv(path)

'''
combined = pd.concat([amd,apple],axis=1)
combined = combined.iloc[:, [j for j, c in enumerate(combined.columns) if j != 6]]

combined = combined[["Price", "Close"]]
combined
'''
```
